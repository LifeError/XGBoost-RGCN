```python
# 定义数据集

# 定义数据特征
x = [1,2,3]

# 定义数据标签
y = [2,4,6]

# 初始化参数W
w = 4

# 定义线性回归模型
def forward(x):
    return x * w

# 定义损失函数
def cost(xs, ys):
    costvalue = 0
    for x, y in zip(xs, ys):
        y_pred = forward(x)
        costvalue += (y - y_pred) ** 2
    return costvalue / len(xs)

# 定义梯度下降函数
def gradient(xs, ys):
    gradvalue = 0
    for x, y in zip(xs, ys):
        gradvalue += 2 * (x * w - y) * x
    return gradvalue / len(xs)

for epoch in range(100):
    cost_val = cost(x, y)

    grad_val = gradient(x, y)

    w -= 0.01 * grad_val

    print(f"训练伦茨：{epoch},损失：{cost_val},更新的w：{w}")
```
